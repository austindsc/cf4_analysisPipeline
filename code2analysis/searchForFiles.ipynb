```python
import os
import glob

def find_missing_files(base_path, run_numbers, file_entries):
  missing_files = []
  foundFiles = []
  for run in run_numbers:
    found = False
    for entry in file_entries:
      # Search recursively using ** pattern
      pattern = os.path.join(base_path, "**", f"*{run}*{entry}*")
      matching_files = glob.glob(pattern, recursive=True)
      if matching_files:
        found = True
        foundFiles.append(run)
        print(f"Found for run {run}: {matching_files}")
        break
    if not found:
      missing_files.append(run)
  return missing_files, foundFiles

# Example usage:
# base_path = "/path/to/start"
# run_numbers = [2022, 2023]
# file_entries = [".root", ".dat"]
# missing = find_missing_files(base_path, run_numbers, file_entries)
# print("Missing runs:", missing)
#
# This will now recursively search ALL subdirectories under base_path for files like:
# - any_folder/xxx2022_test.root
# - deep/nested/folder/data_2022_analysis.root.backup
# - proc/2023_processed.dat
```
