```python
import os
import glob

base = os.environ.get("CF4HOME")

def find_missing_files(base_path, run_numbers, file_entries, verbose = 0):
  missing_files = []
  foundFiles = []
  foundPaths = set()
  for run in run_numbers:
    found = False
    for entry in file_entries:
      # Search recursively using ** pattern
      pattern = os.path.join(base_path, "**", f"*{run}*{entry}*")
      matching_files = glob.glob(pattern, recursive=True)
      if matching_files:
        found = True
        foundFiles.append(run)
        foundPaths.update(os.path.dirname(f) for f in matching_files)
        if verbose > 0:
          print(f"Found for run {run}: {matching_files}")
        break
    if not found:
      missing_files.append(run)
  return missing_files, foundFiles, sorted(foundPaths)

# Example usage:
# base_path = "/path/to/start"
# run_numbers = [2022, 2023]
# file_entries = [".root", ".dat"]
# missing, found, unique_paths = find_missing_files(base_path, run_numbers, file_entries)
# print("Missing runs:", missing)
# print("Found files:", found)
# print("Unique paths:", unique_paths)
#
# This will now recursively search ALL subdirectories under base_path for files like:
# - any_folder/xxx2022_test.root
# - deep/nested/folder/data_2022_analysis.root.backup
# - proc/2023_processed.dat
```

### Checking for Cs LP data

```python
## search for files in Cs run lists
import numpy as np

signalRuns = np.loadtxt(os.path.join(base, "comp2/runLists/CsLists/LPfilter_april_48V.txt"), delimiter=',', usecols=0, dtype=int)
backgroundRuns = np.loadtxt(os.path.join(base, "comp2/runLists/CsLists/LPfilter_april_48V_background.txt"), delimiter=',', usecols=0, dtype=int)

searchFor = signalRuns.tolist() + backgroundRuns.tolist()
```

```python
missing, found, ups = find_missing_files(os.environ.get("CF4HOME"), searchFor, [".bin"])

print("Missing runs:", missing)
print("Found files:", found)
print("Unique paths:", ups)
```

```python
missing, found, ups = find_missing_files(os.environ.get("CF4HOME"), searchFor, [".root"], verbose=1)

print("Missing runs:", missing)
print("Found files:", found)
print("Unique paths:", ups)
```
